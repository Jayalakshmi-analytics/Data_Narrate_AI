```python
# STEP 7: INTERACTIVE DATA CLEANING STUDIO

import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output
from google.colab import files

# Preserve original dataset
df_clean = df.copy()

# Heading
title = widgets.HTML(
    "<h2>🧹 Data Narrate AI - Data Cleaning Studio</h2>"
    "<p>Explore and clean your dataset interactively.</p>"
)

# Select column
column_dropdown = widgets.Dropdown(
    options=list(df_clean.columns),
    description="Column:",
    layout=widgets.Layout(width="80%")
)

# Cleaning options
strategy_dropdown = widgets.Dropdown(
    options=[
        "Leave Unchanged",
        "Fill with Mean",
        "Fill with Median",
        "Fill with Mode",
        "Fill with Custom Value",
        "Drop Rows with Missing Values"
    ],
    description="Action:",
    layout=widgets.Layout(width="80%")
)

# Custom value
custom_value = widgets.Text(
    description="Value:",
    placeholder="Enter replacement value",
    layout=widgets.Layout(width="80%")
)

# Remove constant columns
remove_constant = widgets.Checkbox(
    value=False,
    description="Remove constant-value columns"
)

# Buttons
apply_button = widgets.Button(
    description="Apply Cleaning",
    button_style="success",
    icon="check"
)

preview_button = widgets.Button(
    description="Preview Dataset",
    button_style="info",
    icon="table"
)

csv_button = widgets.Button(
    description="Download CSV",
    button_style="primary",
    icon="download"
)

excel_button = widgets.Button(
    description="Download Excel",
    button_style="primary",
    icon="download"
)

reset_button = widgets.Button(
    description="Reset Dataset",
    button_style="warning"
)

output = widgets.Output()

# Apply cleaning

def apply_cleaning(b):
    global df_clean, cleaning_history

    with output:
        clear_output(wait=True)

        col = column_dropdown.value
        action = strategy_dropdown.value
        before_missing = df_clean[col].isnull().sum()

        try:
            if action == "Fill with Mean":
                if not pd.api.types.is_numeric_dtype(df_clean[col]):
                    print("Mean is available only for numerical columns.")
                else:
                    df_clean[col] = df_clean[col].fillna(df_clean[col].mean())
                    print("Missing values filled using Mean.")

            elif action == "Fill with Median":
                if not pd.api.types.is_numeric_dtype(df_clean[col]):
                    print("Median is available only for numerical columns.")
                else:
                    df_clean[col] = df_clean[col].fillna(df_clean[col].median())
                    print("Missing values filled using Median.")

            elif action == "Fill with Mode":
                mode_values = df_clean[col].mode(dropna=True)
                if len(mode_values) > 0:
                    df_clean[col] = df_clean[col].fillna(mode_values.iloc[0])
                    print("Missing values filled using Mode.")
                else:
                    print("No available mode for this column.")

            elif action == "Fill with Custom Value":
                value = custom_value.value
                if value.strip() == "":
                    print("Please enter a custom value.")
                else:
                    if pd.api.types.is_numeric_dtype(df_clean[col]):
                        value = pd.to_numeric(value)
                    df_clean[col] = df_clean[col].fillna(value)
                    print("Missing values filled with custom value.")

            elif action == "Drop Rows with Missing Values":
                before = len(df_clean)
                df_clean = df_clean.dropna(subset=[col]).copy()
                print(f"Removed {before - len(df_clean)} rows.")

            elif action == "Leave Unchanged":
                print("No changes made.")

            if remove_constant.value:
                constant_cols = [
                    c for c in df_clean.columns
                    if df_clean[c].nunique(dropna=True) <= 1
                ]
                df_clean = df_clean.drop(columns=constant_cols)
                print("Removed constant columns:", constant_cols)

            after_missing = df_clean[col].isnull().sum()

            if before_missing > after_missing:
                new_record = pd.DataFrame([{
                    "Column": col,
                    "Cleaning Method": action,
                    "Values Handled": int(before_missing - after_missing)
                }])

                cleaning_history = pd.concat(
                    [cleaning_history, new_record],
                    ignore_index=True
                )

            print("\nRemaining Rows:", len(df_clean))
            print("Remaining Columns:", len(df_clean.columns))
            print("Missing Cells:", df_clean.isnull().sum().sum())

        except Exception as e:
            print("Cleaning error:", e)

    with output:
        clear_output(wait=True)

        print("CLEANED DATASET PREVIEW")
        print("Rows:", df_clean.shape[0])
        print("Columns:", df_clean.shape[1])

        display(df_clean.head(10))

        print("\nMISSING VALUES REPORT")
        display(df_clean.isnull().sum().to_frame(
            name="Missing Values"
        ))


# Download CSV
def download_csv(b):

    df_clean.to_csv(
        "Data_Narrate_Cleaned.csv",
        index=False
    )

    files.download("Data_Narrate_Cleaned.csv")


# Download Excel
def download_excel(b):

    df_clean.to_excel(
        "Data_Narrate_Cleaned.xlsx",
        index=False
    )

    files.download("Data_Narrate_Cleaned.xlsx")


# Reset original dataset
def reset_dataset(b):
    with output:
        clear_output(wait=True)

        print("CLEANED DATASET PREVIEW")
        print("Total Rows:", df_clean.shape[0])
        print("Total Columns:", df_clean.shape[1])

        print("\nFirst 10 Rows:")
        display(df_clean.head(10))

        print("\nMISSING VALUES REPORT")
        display(df_clean.isnull().sum().to_frame(name="Missing Values"))

        print("\nDataset preview completed!")



    df_clean = df.copy()

    with output:
        clear_output(wait=True)
        print("Dataset successfully reset to original version.")


# Connect buttons
apply_button.on_click(apply_cleaning)
preview_button.on_click(preview_dataset)
csv_button.on_click(download_csv)
excel_button.on_click(download_excel)
reset_button.on_click(reset_dataset)

# Display interactive application
ui = widgets.VBox([
    title,
    widgets.HTML("<h3>1. Select Data Column</h3>"),
    column_dropdown,
    widgets.HTML("<h3>2. Choose Cleaning Method</h3>"),
    strategy_dropdown,
    custom_value,
    remove_constant,
    apply_button,
    widgets.HTML("<hr><h3>3. Dataset Preview</h3>"),
    preview_button,
    output,
    widgets.HTML("<hr><h3>4. Export Cleaned Dataset</h3>"),
    widgets.HBox([csv_button, excel_button, reset_button])
])

display(ui)
```

```python
import gradio as gr
import matplotlib.pyplot as plt
import pandas as pd
import tempfile
import os

# Analyze uploaded dataset
def analyze_dataset(file):

    if file is None:
        return None, None, None, None, None, None, None

    try:
        if file.endswith(".csv"):
            data = pd.read_csv(file)
        else:
            data = pd.read_excel(file)

        rows = len(data)
        columns = len(data.columns)
        missing = int(data.isnull().sum().sum())
        total_cells = data.size
        available = total_cells - missing

        percentage = round(
            (missing / total_cells) * 100, 2
        ) if total_cells else 0

        duplicates = int(data.duplicated().sum())

        missing_report = pd.DataFrame({
            "Column": data.columns,
            "Missing Values": data.isnull().sum().values,
            "Missing Percentage": (
                data.isnull().mean() * 100
            ).round(2).values
        })

        missing_report = missing_report[
            missing_report["Missing Values"] > 0
        ].sort_values(by="Missing Percentage", ascending=False)

        # Missing values chart
        fig1, ax1 = plt.subplots(figsize=(10, 5))

        if not missing_report.empty:
            ax1.barh(
                missing_report["Column"],
                missing_report["Missing Values"]
            )
            ax1.set_title("Missing Values by Column")
            ax1.invert_yaxis()
        else:
            ax1.text(0.5, 0.5, "No Missing Values",
                     ha="center", va="center")
            ax1.set_axis_off()

        fig1.tight_layout()

        # Missing vs available chart
        fig2, ax2 = plt.subplots(figsize=(6, 6))

        if total_cells > 0:
            ax2.pie(
                [missing, available],
                labels=["Missing Data", "Available Data"],
                autopct="%1.1f%%",
                startangle=90
            )

        ax2.set_title("Missing vs Available Data")
        fig2.tight_layout()

        # Gemini AI report
        prompt = f"""
        You are Data Narrate AI.

        Explain this dataset in simple English.

        Rows: {rows}
        Columns: {columns}
        Missing Cells: {missing}
        Missing Percentage: {percentage}%
        Duplicate Rows: {duplicates}

        Column-wise missing values:
        {missing_report.to_string(index=False)}

        Explain the major data quality problems,
        important columns and cleaning recommendations.

        Do not assume why data is missing without evidence.
        """

        response = client.models.generate_content(
            model="gemini-3.5-flash-lite",
            contents=prompt
        )

        report = f"""
DATA NARRATE AI - DATA QUALITY REPORT

Total Rows: {rows}
Total Columns: {columns}
Missing Cells: {missing}
Missing Percentage: {percentage}%
Duplicate Rows: {duplicates}

COLUMN-WISE MISSING VALUES

{missing_report.to_string(index=False)}

AI-GENERATED INSIGHTS

{response.text}
"""

        report_file = tempfile.NamedTemporaryFile(
            mode="w",
            suffix=".txt",
            delete=False,
            encoding="utf-8"
        )

        report_file.write(report)
        report_file.close()

        summary = f"""
## DATA QUALITY REPORT

| Metric | Result |
|---|---:|
| Total Rows | {rows:,} |
| Total Columns | {columns} |
| Missing Cells | {missing:,} |
| Missing Percentage | {percentage}% |
| Duplicate Rows | {duplicates} |

## AI-GENERATED INSIGHTS

{response.text}
"""

        return (
            summary,
            fig1,
            fig2,
            report_file.name,
            gr.Dropdown(choices=list(data.columns),
                        value=data.columns[0],
                        label="Select Column"),
            data,
            data.head(10)
        )

    except Exception as e:
        raise gr.Error(str(e))


# Clean selected column
def clean_dataset(data, column, method):

    if data is None:
        raise gr.Error("Please upload and analyze a dataset first.")

    if not column or column not in data.columns:
        raise gr.Error("Please select a valid column.")

    cleaned = data.copy()

    if method == "Fill with Mean":
        if not pd.api.types.is_numeric_dtype(cleaned[column]):
            raise gr.Error("Mean is only available for numerical columns.")
        cleaned[column] = cleaned[column].fillna(cleaned[column].mean())

    elif method == "Fill with Median":
        if not pd.api.types.is_numeric_dtype(cleaned[column]):
            raise gr.Error("Median is only available for numerical columns.")
        cleaned[column] = cleaned[column].fillna(cleaned[column].median())

    elif method == "Fill with Mode":
        mode_values = cleaned[column].mode(dropna=True)
        if len(mode_values) > 0:
            cleaned[column] = cleaned[column].fillna(mode_values.iloc[0])

    elif method == "Drop Rows with Missing Values":
        cleaned = cleaned.dropna(subset=[column]).copy()

    # Create CSV download
    csv_file = tempfile.NamedTemporaryFile(
        suffix=".csv", delete=False
    )
    csv_file.close()
    cleaned.to_csv(csv_file.name, index=False)

    # Create Excel download
    excel_file = tempfile.NamedTemporaryFile(
        suffix=".xlsx", delete=False
    )
    excel_file.close()
    cleaned.to_excel(excel_file.name, index=False)

    status = f"""
Cleaning completed successfully!

Selected Column: {column}
Cleaning Method: {method}
Remaining Rows: {len(cleaned)}
Remaining Columns: {len(cleaned.columns)}
Remaining Missing Cells: {cleaned.isnull().sum().sum()}
"""

    return cleaned, status, cleaned.head(10), csv_file.name, excel_file.name


# Create Gradio website
with gr.Blocks(title="Data Narrate AI") as app:

    gr.Markdown("# 📊 Data Narrate AI")
    gr.Markdown("### Your AI-Powered Data Analysis Assistant")

    with gr.Tab("Data Analysis"):

        upload_file = gr.File(
            label="Upload CSV or Excel Dataset",
            file_types=[".csv", ".xlsx"],
            type="filepath"
        )

        analyze_button = gr.Button(
            "Analyze Dataset", variant="primary"
        )

        result = gr.Markdown()

        chart1 = gr.Plot(label="Missing Values by Column")
        chart2 = gr.Plot(label="Missing vs Available Data")

        download_report = gr.File(
            label="Download AI Report"
        )

    with gr.Tab("Data Cleaning Studio"):

        gr.Markdown("### 🧹 Clean Your Dataset")

        column_dropdown = gr.Dropdown(
            label="Select Column",
            choices=[]
        )

        strategy_dropdown = gr.Dropdown(
            label="Select Cleaning Method",
            choices=[
                "Fill with Mean",
                "Fill with Median",
                "Fill with Mode",
                "Drop Rows with Missing Values"
            ],
            value="Fill with Median"
        )

        clean_button = gr.Button(
            "Apply Cleaning", variant="primary"
        )

        cleaning_status = gr.Textbox(
            label="Cleaning Result",
            lines=5
        )

        cleaned_preview = gr.Dataframe(
            label="Cleaned Dataset Preview"
        )

        download_csv = gr.File(
            label="Download Cleaned CSV"
        )

        download_excel = gr.File(
            label="Download Cleaned Excel"
        )

    dataset_state = gr.State()

    analyze_button.click(
        fn=analyze_dataset,
        inputs=upload_file,
        outputs=[
            result,
            chart1,
            chart2,
            download_report,
            column_dropdown,
            dataset_state,
            cleaned_preview
        ]
    )

    clean_button.click(
        fn=clean_dataset,
        inputs=[
            dataset_state,
            column_dropdown,
            strategy_dropdown
        ],
        outputs=[
            dataset_state,
            cleaning_status,
            cleaned_preview,
            download_csv,
            download_excel
        ]
    )

print("Data Narrate AI with Data Cleaning Studio is ready!")
```
